```python
import ee
import geemap
import eerepr
ee.Initialize()
eerepr.initialize()
```

```python
Map = geemap.Map()
Map
```

```python
ponto_interesse = ee.Geometry.Point([-122.4439, 37.7538])
print(ponto_interesse)
```

```python
# Seleciona a cena Landsat 8 menos nublada sobre o ponto no período escolhido.
image = (ee.ImageCollection('LANDSAT/LC08/C02/T1_L2')
    .filterBounds(ponto_interesse)
    .filterDate('2016-01-01', '2016-12-01')  # data final exclusiva
    .sort('CLOUD_COVER')
    .first())

# Falso colorido: NIR (SR_B5), vermelho (SR_B4), verde (SR_B3).
# Os valores SR estão armazenados escalados no asset; min/max controlam apenas a exibição.
vis_params = {
    'bands': ['SR_B5', 'SR_B4', 'SR_B3'],
    'min': 0,
    'max': 30000,
    'gamma': 1.2
}

Map.centerObject(ponto_interesse, 12)
Map.addLayer(image, vis_params, 'Landsat 8 — falso colorido SR_B5/SR_B4/SR_B3')
Map
```

```python
# NLCD tem um produto para cada época; seleciona explicitamente o ano de 2019.
# O ponto sozinho tem área zero, então cria-se um buffer de 5 km para o recorte.
nlcd = ee.ImageCollection("USGS/NLCD_RELEASES/2019_REL/NLCD")

Map.addLayer(nlcd,{},"NLCD")
```
